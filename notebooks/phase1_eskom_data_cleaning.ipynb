```python
from datetime import datetime
import pandas as pd

def clean_data(eskom):
    eskom = eskom.fillna({"Original Res Forecast before Lockdown": 0,"Dispatchable Generation": 0, "Residual Demand": 0, "RSA Contracted Demand": 0, 
                          "International Exports": 0, "International Imports": 0, "Thermal Generation": 0, "Nuclear Generation": 0, 
                          "Eskom Gas Generation": 0, "Eskom OCGT Generation": 0, "Hydro Water Generation": 0, "Pumped Water Generation": 0, 
                          "ILS Usage": 0, "Manual Load_Reduction(MLR)": 0, "IOS Excl ILS and MLR": 0, "Dispatchable IPP OCGT": 0, 
                          "Eskom Gas SCO": 0, "Eskom OCGT SCO": 0, "Hydro Water SCO": 0, "Pumped Water SCO Pumping": 0, "Wind": 0, "PV": 0, 
                          "CSP": 0, "Other RE": 0, "Total RE": 0, "Wind Installed Capacity": 0, "PV Installed Capacity": 0, 
                          "CSP Installed Capacity": 0, "Other RE Installed Capacity": 0, "Total RE Installed Capacity": 0})
    
    eskom = eskom.drop(columns=["Original Res Forecast before Lockdown"])
    
    def Date_Time(Date_Time_Hour_Beginning):
        
        date1 = datetime.strptime(Date_Time_Hour_Beginning, "%Y-%m-%d %I:%M:%S %p")
        return date1.strftime("%Y-%m-%d %H:%M:%S")
    
    eskom.insert(1, "Date Time", eskom.apply(lambda row : Date_Time(row["Date Time Hour Beginning"]), axis=1))
    eskom = eskom.drop(columns=["Date Time Hour Beginning"])
    eskom.columns = eskom.columns.str.replace(" ", "_")
    eskom.columns = eskom.columns.str.lower()
    return eskom

eskom = pd.read_csv("ESK18239.csv")
eskom_clean = clean_data(eskom.copy())
eskom_clean.to_csv("eskom_clean.csv", index = False)
print(eskom_clean.head())
```
