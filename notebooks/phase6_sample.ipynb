## Save a Sample for the Dashboard

```python
import pandas as pd
import numpy as np

df = pd.read_csv("settlements_saws_scored.csv", low_memory = False)
 
# fill all nulls with 0 for numeric columns 
numeric_cols = df.select_dtypes(np.number).columns 
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors = "coerce").fillna(0)
 
# fill string nulls with unkown
string_cols = df.select_dtypes(include = "object").columns 
df[string_cols] = df[string_cols].fillna("Unknown")

print("After cleaning:")
print(df.isnull().sum().sum(), "nulls remaining")

# sample 6 000 rows 
df.sample(n = 6000, random_state = 42).to_excel("settlements_saws_scored_sample.xlsx", index = False)
print("Saved sampled file")
```

```python
files = {
    "solar_farm_recommendations_final.csv": "solar_farm_recommendations_final.xlsx",
    "optimisation_summary.csv": "optimisation_summary.xlsx"
}

for csv_file, excel_file in files.items():
    csv = pd.read_csv(csv_file, low_memory = False)
    
    # only convert numeric columns to float
    for col in csv.columns:
        if csv[col].dtype in ["float", "int64"]:
            csv[col] = pd.to_numeric(csv[col], errors = "coerce").fillna(0)
        
    # verify string columns
    string_cols = csv.select_dtypes(include = "object").columns.tolist()
    print(csv_file)
    print(f"String columns preserved: {string_cols}")
    print(f"Numeric columns: {csv.select_dtypes(include = np.number).columns.tolist()}")
    csv.to_excel(excel_file, index = False)
    print(f"Saved: {excel_file}")
```
